## 7. 🔍 IDENTIFICAÇÃO DE PROBLEMAS - Valores Nulos

```python
print('='*70)
print('ANÁLISE DE VALORES NULOS')
print('='*70)

# Contagem de nulos por coluna
nulos = df.isnull().sum()
percentual_nulos = (df.isnull().sum() / len(df)) * 100

# Criar DataFrame com a análise
analise_nulos = pd.DataFrame({
    'Coluna': df.columns,
    'Qtd_Nulos': nulos.values,
    'Percentual': percentual_nulos.values
})

# Ordenar por quantidade de nulos (decrescente)
analise_nulos = analise_nulos[analise_nulos['Qtd_Nulos'] > 0].sort_values('Qtd_Nulos', ascending=False)

if len(analise_nulos) == 0:
    print('✅ EXCELENTE! Não há valores nulos no dataset.')
else:
    print(f'⚠️ Encontrados valores nulos em {len(analise_nulos)} coluna(s):\n')
    for idx, row in analise_nulos.iterrows():
        print(f"  • {row['Coluna']:30s} → {row['Qtd_Nulos']:5.0f} nulos ({row['Percentual']:5.2f}%)")
    
    print(f"\n📊 Total de células com nulo: {nulos.sum():,}")
    print(f"📊 Linhas com pelo menos 1 nulo: {df.isnull().any(axis=1).sum():,}")
```

```python
# Visualizar nulos (se houver)
if len(analise_nulos) > 0:
    plt.figure(figsize=(12, 6))
    plt.barh(analise_nulos['Coluna'], analise_nulos['Percentual'])
    plt.xlabel('Percentual de Valores Nulos (%)')
    plt.title('Distribuição de Valores Nulos por Coluna')
    plt.tight_layout()
    plt.show()
```

## 8. 🔍 IDENTIFICAÇÃO DE PROBLEMAS - Linhas Duplicadas

```python
print('='*70)
print('ANÁLISE DE LINHAS DUPLICADAS')
print('='*70)

# Verificar duplicatas completas (todas as colunas)
duplicatas_completas = df.duplicated().sum()
percentual_duplicatas = (duplicatas_completas / len(df)) * 100

print(f'Total de registros: {len(df):,}')
print(f'Linhas duplicadas (100%): {duplicatas_completas:,} ({percentual_duplicatas:.2f}%)')

if duplicatas_completas > 0:
    print(f'\n⚠️ Encontradas {duplicatas_completas} linha(s) totalmente duplicada(s)!')
    print('\n📋 Exemplos de linhas duplicadas:')
    display(df[df.duplicated(keep=False)].sort_values(by=df.columns.tolist()).head(10))
else:
    print('\n✅ EXCELENTE! Não há linhas duplicadas no dataset.')
```
