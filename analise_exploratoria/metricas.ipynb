---
### 1. Imports e Configurações

```python
import pandas as pd
import numpy as np
import warnings

# Configurações
warnings.filterwarnings('ignore')
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)

print("✅ Bibliotecas importadas com sucesso!")
```

---
### 2. Carregamento dos Dados

```python
# Caminho do arquivo
caminho_arquivo = r'C:\Users\Levi Brabo\Desktop\projects\analise_comercial\data\RelatorioProcessos_BA_MAR_ate_SET2026.xlsx'

# Carregamento
print("⏳ Carregando dados...")
df_original = pd.read_excel(caminho_arquivo)

print(f"✅ Dados carregados com sucesso!")
print(f"📊 Dimensões originais: {df_original.shape[0]} linhas x {df_original.shape[1]} colunas")
```

---
### 3. Limpeza Automática de Dados

```python
# Cópia para trabalhar
df = df_original.copy()

# Contadores
linhas_originais = len(df)

print("🧹 INICIANDO LIMPEZA...\n")

# 1. Remover linhas completamente vazias
df = df.dropna(how='all')
vazias_removidas = linhas_originais - len(df)
print(f"✓ Linhas completamente vazias removidas: {vazias_removidas}")

# 2. Remover duplicatas
linhas_antes_duplicatas = len(df)
df = df.drop_duplicates()
duplicatas_removidas = linhas_antes_duplicatas - len(df)
print(f"✓ Duplicatas removidas: {duplicatas_removidas}")

# Resumo final
print("\n" + "="*50)
print("📋 RESUMO DA LIMPEZA")
print("="*50)
print(f"Linhas originais:        {linhas_originais:,}")
print(f"Linhas vazias removidas: {vazias_removidas:,}")
print(f"Duplicatas removidas:    {duplicatas_removidas:,}")
print(f"Linhas finais:           {len(df):,}")
print(f"Total removido:          {linhas_originais - len(df):,}")
print("="*50)
```

---
### 4. Informações do DataFrame Limpo

```python
print("📊 INFORMAÇÕES DO DATASET LIMPO\n")

# Dimensões
print(f"Dimensões: {df.shape[0]:,} linhas x {df.shape[1]} colunas\n")

# Tipos de dados
print("Tipos de dados:")
print(df.dtypes)
print("\n" + "="*50)
```

---
### 7. Análise de Valores Nulos

```python
# Contagem de nulos por coluna
nulos = df.isnull().sum()
nulos_pct = (df.isnull().sum() / len(df) * 100).round(2)

analise_nulos = pd.DataFrame({
    'Coluna': df.columns,
    'Nulos': nulos.values,
    '% Nulos': nulos_pct.values
})

# Filtrar apenas colunas com nulos
analise_nulos = analise_nulos[analise_nulos['Nulos'] > 0].sort_values('Nulos', ascending=False)

print("🔍 ANÁLISE DE VALORES NULOS:\n")
if len(analise_nulos) > 0:
    display(analise_nulos)
else:
    print("✅ Nenhum valor nulo encontrado!")
```
